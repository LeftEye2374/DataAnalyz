```python
import pandas as pd
```

```python
df = pd.read_csv('data/train.csv')
df_copy = df.copy()
```

Проверка пропусков


```python
df_copy.isnull().sum()
```

Заполняем нули в Age медианной

```python
df_copy['Age'] = df_copy['Age'].fillna(df_copy['Age'].median())
```

```python
df_copy.isnull().sum()
```

Заполнил строковые данные в графе Пол на 0 и 1

```python
df_copy['Sex'] = df_copy['Sex'].map({'male': 0, 'female': 1})
df_copy = pd.get_dummies(df_copy, columns=['Embarked'], drop_first=True)
```

```python
df_copy = df_copy.drop(columns=['Cabin', 'Name', 'Ticket', 'PassengerId'])
```

```python
df_copy.dtypes
```

```python
df_copy.isnull().sum()
```

Подключил библиотеки для обучения

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
x = df_copy.drop(columns=['Survived'])
y = df_copy['Survived']

x.columns.tolist()
x.dtypes

x_train, x_val, y_train, y_val = train_test_split(x, y, test_size=0.2, random_state=42)

model = LogisticRegression(max_iter = 1000)
model.fit(x_train, y_train)
```

Метрики и оценка модели

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix

y_pred = model.predict(x_val)

print("Accuracy:", accuracy_score(y_val, y_pred))
print("Precision:", precision_score(y_val, y_pred))
print("Recall:", recall_score(y_val, y_pred))
print(confusion_matrix(y_val, y_pred))
```

Рандом форест


```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(x_train, y_train)

y_pred_rf = rf_model.predict(x_val)
print("Accuracy:", accuracy_score(y_val, y_pred_rf))
print("Precision:", precision_score(y_val, y_pred_rf))
print("Recall:", recall_score(y_val, y_pred_rf))
print(confusion_matrix(y_val, y_pred_rf))
```
