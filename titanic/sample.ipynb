```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Read data from csv files

```python
gender_set = pd.read_csv("data/gender_submission.csv")
test_set = pd.read_csv("data/test.csv")
train_set = pd.read_csv("data/train.csv")
```

```python
test_set.head()
```

```python
test_set.columns
```

```python
test_set.boxplot(column="Age", by="Sex")
plt.xlabel("Sex")
plt.ylabel("Age")
plt.title("Age distribution by sex")
plt.show()
```

```python
merged = test_set.merge(gender_set, on="PassengerId")
```

```python
merged.groupby('Pclass')["Survived"].mean().plot(kind='bar')
plt.xlabel("Pclass")
plt.ylabel("Survival rate")
plt.title("Survival rate by passenger class")
plt.show()
```

```python
train_set['Age'].hist(bins=30)
plt.xlabel("Age")
plt.ylabel("Count")
plt.title("Age distribution by age")
plt.show()
```
